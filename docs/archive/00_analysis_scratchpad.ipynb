* External data and industry analysis confirm that 2014 was indeed the commencement of the "accelerating growth pattern" that defines the current era.

### Math
**First Derivative**

In calculus, the "first derivative" measures the speed of change.  


Comparing how fast launches were increasing in the old model (legacy linear) versus how fast they are increasing in the new model (modern exponential).

We use Brent's Method to find the exact year where the modern exponential growth rate officially surpassed the legacy linear growth rate.

Calculating the slopes lets us compare how fast launches were increasing in the old model versus how fast they were increasing in the new one.

### **[Brent's Method](https://en.wikipedia.org/wiki/Brent%27s_method)**


Using Brent's Method to find the exact moment where the two speeds (slopes) became equal (2014) and we interpret this as the year the transition occurs (root), because the models velocity officially surpasses the legacy trend during that calendar year.

- growth models
  * two distinct mathematical functions help us characterize the two different eras of space flight:
    * steady state (linear growth): $y = mx + b$  
      * $m$ is the constant slope (legacy launch cadence) and $b$ is the intercept.
    * new space (exponential growth): $y = a \cdot e^{b(x-c)}$  
      * $a$ is the scale, $b$ is the growth rate, and $x-c$ is the time elapsed since the baseline year ($c$).
- acceleration math
  * exponential slope (($f'(x)$)): $\frac{d}{dx}[a \cdot e^{b(x-c)}] = a \cdot b \cdot e^{b(x-c)}$
    * calculates the exact slope of the exponential curve at any given year  
- root finding solver
  * root: exact year where the  exponential growth rate ($f'(x)$) officially surpassed the legacy cadence

### Possibly Expansion Path - Prompt Response

**If I wanted to start building out a vue, or react web ui that interacts with my notebooks, charts, python code, or database files (maybe an API or something) to expand my initial project, how would I best approach it?**

Given your repo as it exists now, the cleanest path is to treat the notebooks as your research layer and build a small Python API in front of your real data products: `data/clean/orbital_debris.db`, `data/clean/results/*.parquet`, and any reusable logic from [notebooks/utility.py](d:/repos/orbital-debris-assessment/notebooks/utility.py) or notebook cells. Then put a React or Vue app on top of that API.

I would not make the web UI talk directly to notebooks. Notebooks are great for exploration and one-off analysis, but they get awkward as an app backend. The better pattern is:

1. Extract reusable Python logic from notebooks into regular modules like `src/orbital_debris/queries.py`, `metrics.py`, and `services.py`.
2. Add a lightweight API layer, preferably `FastAPI`, because it fits your Python stack and works well with SQLite, pandas, and parquet.
3. Build the frontend separately in `frontend/` with either React or Vue.
4. Have the frontend request JSON from the API and render charts/tables interactively.

For your project specifically, that probably means endpoints like:

- `GET /api/summary`
- `GET /api/high-risk-objects`
- `GET /api/launch-trends`
- `GET /api/top-owners`
- `GET /api/chart/{name}`

You already have precomputed parquet outputs, which is a big advantage. Early on, your API can simply read those parquet files and return them as JSON instead of recomputing heavy notebook logic on every request. That gives you a fast, stable first version.

If you want the most practical stack choice, I’d suggest React + FastAPI unless you already prefer Vue. React has slightly more examples and ecosystem support for data-heavy dashboards, but Vue would work just as well. The important decision is less “React vs Vue” and more “separate frontend app + Python API + extracted notebook logic.”

A good build order would be:

- Phase 1: Create `backend/` or `api/` with FastAPI and one endpoint reading from `orbital_debris.db`.
- Phase 2: Add a React or Vue dashboard with 2-3 pages: overview, trends, high-risk objects.
- Phase 3: Move notebook SQL/query code into shared Python modules so both notebooks and API use the same logic.
- Phase 4: Add filters, drill-downs, and chart interactivity.
- Phase 5: Optionally deploy with the frontend on Vercel/Netlify and the API on Render/Fly.io, or keep both local for portfolio/demo use.

A simple target structure could look like:

```text
orbital-debris-assessment/
  api/
    main.py
    routes/
    services/
    schemas/
  src/
    orbital_debris/
      queries.py
      loaders.py
      metrics.py
  frontend/
    src/
      pages/
      components/
      lib/
```

A few practical rules will save you pain:

- Keep notebooks for exploration and storytelling, not app logic.
- Put all repeatable business/query logic in Python modules.
- Let the API own database/parquet access.
- Let the frontend own presentation, filters, and interactions.
- Start with read-only features first; you probably don’t need writes yet.

If you want the fastest “first UI” with less engineering, `Streamlit` is also a valid stepping stone. But if your goal is a more professional expandable web app, `FastAPI + React` or `FastAPI + Vue` is the better long-term foundation.

**How is django these days?**

Django is still very good in 2026. It’s mature, actively maintained, and still one of the best choices if you want a Python-first web app with a real data model, authentication, admin tooling, and room to grow. As of April 7, 2026, Django had current security releases for `6.0.4` and `5.2.13`, and `4.2` just reached end of extended support, so it’s clearly still healthy and moving. Sources: https://www.djangoproject.com/weblog/2026/apr/07/security-releases/ and https://docs.djangoproject.com/en/6.0/internals/release-process/

The short version is:
- Django is excellent for “full app” work.
- FastAPI is usually nicer for “thin API in front of notebooks/data”.
- Django is no longer “old-fashioned,” but it is still more opinionated and heavier than FastAPI.

What’s good about Django now:
- Strong ORM, migrations, auth, forms, sessions, and admin out of the box.
- Great if you may eventually want users, saved dashboards, permissions, uploads, or internal tooling.
- Supports ASGI and async views, though the ORM is still not fully async-native yet. Official async docs say async support exists, but parts like the ORM still require sync bridging in places: https://docs.djangoproject.com/en/6.0/topics/async/

For your project, I’d frame it like this:
- If you want a polished data platform that might grow into accounts, saved analyses, admin-managed metadata, or a serious portfolio app, Django is a strong choice.
- If you mainly want a React/Vue frontend that reads SQLite/parquet/query results through a lightweight API, `FastAPI` is probably the simpler start.

My honest recommendation for this repo:
- `FastAPI + React/Vue` if you want the fastest clean path from notebooks to web UI.
- `Django + Django REST Framework + React/Vue` if you expect the app to become a fuller product with users, auth, admin, and more backend structure.

A nice middle-ground is Django for backend plus a separate frontend. You don’t have to use Django templates if you don’t want to.
