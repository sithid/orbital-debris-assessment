## Primary Question 3: Who Leaves the Most "Zombie" Satellites in Space?

**Which ownership or operator groups are responsible for the most non-working "zombie" satellites?**

When satellites stop working but stay in orbit, they become "zombies." These dead satellites can’t move out of the way or avoid crashes. They add to the crowding in space and increase the risk for working satellites. Knowing which countries or companies have left the most zombie satellites helps us see who could make the biggest difference in cleaning up.
