# Orbital Debris Assessment: Story Questions

Understanding the accelerating risks in low Earth orbit requires more than just counting objects—it demands a structured, query-driven approach that reveals when and how orbital growth shifted from a steady, linear trend to a new era of rapid expansion. This notebook builds on a master list of satellites and debris, normalized into a relational database, to answer targeted research questions about the timing of this growth transition and the concentration of high-risk objects by orbit class, altitude band, and owner/operator. Each section below presents a focused analysis, using SQL-backed queries and visual storytelling to map the evolving landscape of orbital congestion and collision risk.

### Primary Questions
1. At what year did orbital growth decouple from a historical linear trend and transition into an exponential trajectory?
2. How are high-risk objects distributed across orbit classes and altitude bands, especially in 400–600 km LEO?
3. Which ownership/operator groups contribute most to non-operational "zombie" payload concentrations?

**Primary Question 1: When Did Orbital Growth Take Off?**

**At what year did orbital growth shift from a slow, steady pace to rapid, exponential growth?**

All evidence points to a single turning point: **2014**. Before then, space activity grew gradually. After 2014, everything accelerates sharply—especially payload launches. The charts below walk through the story, step by step.

---

**How Many Objects Are Launched Each Year?**

- **Payloads per Year:**
  ![Payloads per Year](../charts/payloads_per_year_exp.png)
  > Payload launches were steady and modest for decades, then surged dramatically after 2015. The recent spike is driven by commercial megaconstellations, reusable rockets, and the miniaturization of satellites—enabling thousands of payloads to be launched each year. This marks a structural shift: space access has moved from rare, government-led missions to frequent, high-volume commercial operations.

- **Rocket Bodies per Year:**
  ![Rocket Bodies per Year](../charts/rocket_bodies_per_year_exp.png)
  > Rocket bodies are a direct measure of launch activity and the hardware left behind in orbit. Launch rates ramped up quickly in the early space age, peaked in the 1980s, then declined as rockets became more efficient and disposal practices improved. In recent years, launch rates have climbed again—driven by commercial providers and satellite constellations—though modern rockets often use better disposal or reusability. Tracking rocket bodies helps us understand both launch trends and ongoing debris risks.

- **Launch Cohorts:**
  ![Launch Cohort Analysis](../charts/launch_cohort_analysis_exp.png)
  > This chart reveals the dramatic transformation of space activity over time. Early years saw gradual growth and debris-heavy launches, but since 2019, the pace has shifted to explosive, exponential expansion—driven by megaconstellations like SpaceX and OneWeb. The recent bars show massive batches of payloads launched together, with the in-orbit population (green line) rising steeply. This new era is defined by frequent, large-scale deployments of satellites, not debris, fundamentally changing the makeup and congestion of Earth orbit. The sharp rise in objects still in orbit highlights both the technological leap and the growing challenge of space sustainability.

**What Changed in How We Launch?**

Before the split, launches produced more debris and fewer payloads, with rocket bodies present in numbers similar to payloads. After 2014, payload launches surged, while debris and rocket bodies per launch dropped sharply. New practices—like rideshare missions and batch deployments—helped increase payloads and reduce debris.

**Primary Question 1 Conclusion**

The evidence is clear: orbital growth shifted from slow to exponential in 2014. The combination of commercial megaconstellations, reusable rockets, small-satellite technology, and lower launch costs created a new era of high-frequency, high-volume access to space. Modern launches are cleaner and more productive, with far more satellites and less debris. This is the new normal for Earth orbit.

**Primary Question 2: Where Are the Most Dangerous Objects in Orbit?**

**At what altitudes and in which orbit classes are high-risk objects most concentrated, especially in the crowded 400–600 km LEO region?**

High-risk objects are the largest, most energetic things in orbit—massive dead satellites and spent rocket bodies that could cause the most damage if they break up or collide. Only a small fraction of all objects are truly high-risk, but they account for most of the collision danger.

---

**How Do Orbit Class and Altitude Combine?**

- LEO dominates every altitude band below 2,000 km.
- The 400–600 km band contains thousands of LEO high-risk objects and hundreds of “unknown-class” objects (often old debris or unclassified payloads).
- Higher orbits (MEO, GEO) have fewer high-risk objects, but those present are often very massive and long-lived.

![High-Risk Object Distribution by Altitude Band and Orbit Class](../charts/high_risk_object_distribution_alt_bands_exp.png)

> LEO is the main stage for collision risk because it’s packed with satellites and debris moving fast and crossing paths often.

---

**Which Orbit Classes Have the Most High-Risk Objects?**

- LEO: ~85% of all high-risk objects
- MEO: ~8%
- GEO: ~4%
- Elliptical: ~3%
- Unknown: <1%

![High-Risk Objects by Orbit Class](../charts/hr_orbit_class_exp.png)
![High-Risk Objects by Orbit Class (Pie)](../charts/high_risk_pie_exp.png)

> LEO is overwhelmingly the center of risk. MEO and GEO matter too, but they’re secondary.

---

**Where Do the Most Dangerous Objects Cluster?**

- High-risk objects aren’t spread out evenly—they cluster in certain altitude bands and orbit classes, creating “hotspots” for risk.
- The 400–600 km band in Low Earth Orbit (LEO) is the most crowded and active, with lots of launches, decays, and close calls.

![High-Risk Objects by Altitude Band](../charts/high_risk_banded_exp.png)

> The 400–600 km band is the “downtown traffic jam” of space—busy, crowded, and full of large objects that can cause major damage.

---

**How Massive Are the Riskiest Objects?**

- Most high-risk objects are in LEO, but the heaviest are spread across multiple orbit classes.
- A few extremely massive objects (10,000–450,000 kg) represent a huge risk and are top targets for removal and close monitoring.

![High-Risk Mass by Orbit Class (Strip)](../charts/high_risk_mass_strip_by_orbit_class.png)
![High-Risk Mass by Orbit Class (Histogram)](../charts/high_risk_mass_hist_by_orbit_class.png)

---

**What About Debris Clouds?**

- Most debris is made up of much smaller fragments, but their numbers and clustering create “debris clouds” that raise the risk for everyone.
- Debris creation isn’t steady—big events like breakups or collisions can suddenly multiply the number of fragments, making space more hazardous.

![Debris Objects Added per Year](../charts/debris_per_year_exp.png)

---

**How Has the Mix of Objects Changed Over Time?**

- Debris spikes in certain years (major breakups)
- Payloads surge after 2018
- Rocket bodies remain steady
- The mix highlights how both new satellites and persistent debris contribute to congestion

![Object Type Distribution Over Time (Bar Chart)](../charts/object_type_distribution_bar_exp.png)

---

**Primary Question 2 Conclusion**

High-risk objects are packed into the 400–600 km LEO band and in LEO overall, making these areas the top priority for risk reduction. But the huge number of smaller debris fragments also matters—a crowded, cluttered space environment means more chances for accidents. Both big, dangerous objects and dense debris clouds need to be managed to keep space safe.

**Which organizations or operator groups are responsible for the most non-working "zombie" satellites?**

They add to the crowding in space and increase the risk for working satellites. Knowing which organizations or operators have left the most zombie satellites helps us see who could make the biggest difference in cleaning up.

# Story Conclusion

This analysis reveals a clear turning point in the history of Earth orbit: since 2014, space has entered an era of exponential growth, driven by commercial megaconstellations, reusable rockets, and the miniaturization of satellites. The most dangerous objects—large, high-energy satellites and rocket bodies—are overwhelmingly concentrated in the 400–600 km band of Low Earth Orbit, creating a crowded and collision-prone hotspot. Responsibility for non-operational “zombie” satellites is highly concentrated among a few organizations and operators, while a handful of leading groups maintain the largest active fleets.

Together, these findings show that orbital congestion and collision risk are not evenly distributed—they are shaped by recent technological shifts, concentrated in specific regions, and driven by the practices of a small number of key players. Addressing these challenges will require targeted action: better end-of-life practices, stricter disposal requirements, and focused cleanup efforts among the biggest contributors. By prioritizing accountability and remediation, the space community can make meaningful progress toward a safer, more sustainable orbital environment for all.
