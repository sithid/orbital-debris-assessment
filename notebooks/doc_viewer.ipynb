# Markdown Reader

This notebook serves as an easy way to view my markdown data dictionarys.  With the continued use of markdown during the last 4 or so Code:You pathways, ive found myself completely attached to the hip with markdown. I find it extremely flexible and great for so many things.  Now that I know how to couple markdown with python and Jupyter Notebooks, markdown is my preferred format for documentation and other such things. My one singluar complaint with markdown is that it is a bit of a pain to actually view if you are not viewing it on github, within preview mode using vscode, or another such plugin.  This utility notebook allows anybody reviewing the documents to be able to easily view the data dictionarys as I intended them to be seen.

Usage: Run the code cell and enter 1 - 4, depending on which schema you are interested in. The code cell will yoink the markdown source from the md file, and display it as notebook markdown using Markdown(). Its fairly simple and easy enough to understand.

I like to run this with the database schema selected while im writing my queries.

```python
# https://publish.obsidian.md/multiversecurriculum/Curriculum/Learn+to+Code+Python/Reading+a+directory+of+files+in+Python
# select and display a markdown data dictionary from docs/schema.
from IPython.display import display, Markdown
import os

# List all .md files in the docs/schema directory
schema_dir = '../docs/schema'
md_files = [f for f in os.listdir(schema_dir) if f.endswith('.md')]

# Display available files for user selection
print('Available data dictionaries:')
for i, fname in enumerate(md_files):
    print(f'{i+1}: {fname}')

# User input to select file
idx = int(input('Enter the number of the file to display: ')) - 1
md_path = os.path.join(schema_dir, md_files[idx])

# Read and display the selected Markdown file
with open(md_path, 'r', encoding='utf-8') as f:
    md_content = f.read()
    
display(Markdown(md_content))
```
