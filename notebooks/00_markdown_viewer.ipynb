# Markdown Viewer

This notebook provides a simple, interactive way to browse and view all Markdown documentation files in the `docs/` folder and its subfolders. It is especially useful for quickly referencing data dictionaries, schema notes, and other project documentation while working in Jupyter.

**How to use:**
- Run the code cell below.
- Select any Markdown file from the dropdown menu.
- Click **Display** to render the file as notebook-formatted Markdown.

**Why this notebook?**
- Keeps all my project documentation easily accessible in one place.
- No need to switch to VS Code preview, GitHub, or external tools to read Markdown.
- Designed for convenience during analysis, query writing, and review.

```python
# https://ipywidgets.readthedocs.io/en/stable/
# https://ipywidgets.readthedocs.io/en/stable/examples/Widget%20List.html

# Use a dropdown and button to select and display any markdown file in ./docs or its subfolders, avoiding infinite loop.
from IPython.display import display, Markdown
import os
import ipywidgets as widgets

# find all .md files in ./docs and subfolders
documents = '../docs'
md_files = []
for root, dirs, files in os.walk(documents):
    for f in files:
        if f.endswith('.md'):
            rel_path = os.path.relpath(os.path.join(root, f), documents)
            md_files.append(rel_path)

# Create dropdown, button, and output widgets, with no default file selected 
dropdown = widgets.Dropdown(
    options=[('Select a file...', None)] + [(f, f) for f in md_files],
    layout=widgets.Layout(width='50%'),
    value=None
)

# I think i like these jupiter notebook widgets better than the other python GUI libraries
# ive used in the past. They are simple and work well enough.
button = widgets.Button(description='Display', button_style='primary')
output = widgets.Output()

def on_button_clicked(b):
    selected = dropdown.value
    if selected is not None:
        # construct the full path to the selected markdown file
        md_path = os.path.join(documents, selected)
        
        # read the file to end
        with open(md_path, 'r', encoding='utf-8') as f:
            md_content = f.read()
        
        # display the markdown content in the output widget
        with output:
            output.clear_output()
            display(Markdown(md_content))

button.on_click(on_button_clicked)
display(widgets.HBox([button, dropdown]), output)
```
